Data Cleaning

```python
# Remove the unnecessary index column
df.drop("Unnamed: 0", axis=1, inplace=True)

# Encode target variable
df["y"] = df["y"].map({"M": 0, "B": 1})

# Verify the dataset
df.head()
```

Feature Scaling using StandardScaler

```python
# Features and Target
X = df.drop("y", axis=1)
y = df["y"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

Apply Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Scale all feature values
X_scaled = scaler.fit_transform(X)

# Display first five scaled rows
print(X_scaled[:5])
```

Train-Test Split Analysis


```python
from sklearn.model_selection import train_test_split

# 80:20 Train-Test Split
X_train_80, X_test_80, y_train_80, y_test_80 = train_test_split(
    X_scaled,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Features:", X_train_80.shape)
print("Testing Features :", X_test_80.shape)
print("Training Target  :", y_train_80.shape)
print("Testing Target   :", y_test_80.shape)
```

Repeat with 70:30 Split

```python
# 70:30 Train-Test Split
X_train_70, X_test_70, y_train_70, y_test_70 = train_test_split(
    X_scaled,
    y,
    test_size=0.30,
    random_state=42
)

print("Training Features:", X_train_70.shape)
print("Testing Features :", X_test_70.shape)
print("Training Target  :", y_train_70.shape)
print("Testing Target   :", y_test_70.shape)
```

Repeat with 90:10 Split

```python
# 90:10 Train-Test Split
X_train_90, X_test_90, y_train_90, y_test_90 = train_test_split(
    X_scaled,
    y,
    test_size=0.10,
    random_state=42
)

print("Training Features:", X_train_90.shape)
print("Testing Features :", X_test_90.shape)
print("Training Target  :", y_train_90.shape)
print("Testing Target   :", y_test_90.shape)
```

KNN Model with Heuristic K Selection

```python
import math

# Number of training samples
n = len(X_train_80)

# Heuristic K value
k = int(math.sqrt(n))

# Ensure K is odd
if k % 2 == 0:
    k += 1

print("Training Samples:", n)
print("Heuristic K Value:", k)
```

Train the KNN Model using Heuristic K

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Create KNN model with heuristic K
knn = KNeighborsClassifier(n_neighbors=k)

# Train the model
knn.fit(X_train_80, y_train_80)

# Predict on test data
y_pred = knn.predict(X_test_80)

# Calculate accuracy
accuracy = accuracy_score(y_test_80, y_pred)

print("Accuracy:", accuracy)
```

Experiment with Nearby K Values (K ± 5)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

k_values = list(range(16, 27))   # K = 16 to 26
accuracies = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_80, y_train_80)

    y_pred = model.predict(X_test_80)

    acc = accuracy_score(y_test_80, y_pred)
    accuracies.append(acc)

    print(f"K = {k} --> Accuracy = {acc:.4f}")
```

Plot Accuracy vs K

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(k_values, accuracies, marker='o')
plt.title("Accuracy vs K")
plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Accuracy")
plt.xticks(k_values)
plt.grid(True)

plt.show()
```

Find the Best K

```python
best_accuracy = max(accuracies)
best_k = k_values[accuracies.index(best_accuracy)]

print("Best K:", best_k)
print("Best Accuracy:", best_accuracy)
```

Reduce the Data to 2 Dimensions (PCA)

```python
from sklearn.decomposition import PCA

# Reduce the scaled features to 2 principal components
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original Shape :", X_scaled.shape)
print("Reduced Shape  :", X_pca.shape)
```

Split the PCA Data

```python
from sklearn.model_selection import train_test_split

# Split the PCA-transformed data
X_train_pca, X_test_pca, y_train_pca, y_test_pca = train_test_split(
    X_pca,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Shape:", X_train_pca.shape)
print("Testing Shape :", X_test_pca.shape)
```

Function to Plot Decision Boundary



```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn.neighbors import KNeighborsClassifier

def plot_decision_boundary(k):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_pca, y_train_pca)

    x_min, x_max = X_train_pca[:, 0].min() - 1, X_train_pca[:, 0].max() + 1
    y_min, y_max = X_train_pca[:, 1].min() - 1, X_train_pca[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, 0.1),
        np.arange(y_min, y_max, 0.1)
    )

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(7,5))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap=ListedColormap(("red", "green")))

    plt.scatter(
        X_train_pca[:, 0],
        X_train_pca[:, 1],
        c=y_train_pca,
        cmap=ListedColormap(("red", "green")),
        edgecolor="k"
    )

    plt.title(f"KNN Decision Boundary (K={k})")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.show()
```

Plot the Required Graphs

```python
plot_decision_boundary(1)
plot_decision_boundary(5)
plot_decision_boundary(10)
plot_decision_boundary(20)
```

Cross Validation

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

k_values = range(16, 27)

cv_scores = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)

    scores = cross_val_score(
        knn,
        X_scaled,
        y,
        cv=5,
        scoring="accuracy"
    )

    cv_scores.append(scores.mean())

    print(f"K = {k} --> Mean Accuracy = {scores.mean():.4f}")
```

Plot Cross Validation Accuracy

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(k_values, cv_scores, marker='o')

plt.title("5-Fold Cross Validation Accuracy vs K")
plt.xlabel("K Value")
plt.ylabel("Mean Accuracy")

plt.xticks(k_values)
plt.grid(True)

plt.show()
```

Best K using Cross Validation

```python
best_cv_accuracy = max(cv_scores)
best_cv_k = list(k_values)[cv_scores.index(best_cv_accuracy)]

print("Best K (Cross Validation):", best_cv_k)
print("Best Mean Accuracy:", best_cv_accuracy)
```
