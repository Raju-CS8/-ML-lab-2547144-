Create the XOR Dataset

```python
import numpy as np

# XOR input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=np.float32)

# XOR output
y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=np.float32)

print("Input X:")
print(X)

print("\nOutput y:")
print(y)
```

Build the MLP using Keras

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
```

```python
from tensorflow import keras
from tensorflow.keras import layers

# Create the MLP model
model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(4, activation='tanh'),
    layers.Dense(1, activation='sigmoid')
])

# Display model architecture
model.summary()
```

Compile the Keras model

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

Train the Keras MLP

```python
history = model.fit(
    X,
    y,
    epochs=1000,
    verbose=0
)

print("Model training completed!")
```

Predict the XOR outputs

```python
predictions = model.predict(X, verbose=0)

print("Predicted probabilities:")
print(predictions)

print("\nPredicted classes:")
print((predictions >= 0.5).astype(int))

print("\nActual outputs:")
print(y.astype(int))
```

Calculate Accuracy

```python
from sklearn.metrics import accuracy_score

predicted_classes = (predictions >= 0.5).astype(int)

accuracy = accuracy_score(y, predicted_classes)

print("Final Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

Build the MLP using PyTorch

```python
import torch

print("PyTorch version:", torch.__version__)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Convert XOR data to PyTorch tensors
X_torch = torch.tensor(X, dtype=torch.float32)
y_torch = torch.tensor(y, dtype=torch.float32)

# Define the MLP
class XOR_MLP(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.network = nn.Sequential(
            nn.Linear(2, 4),
            nn.Tanh(),
            nn.Linear(4, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x)

# Create the model
torch_model = XOR_MLP()

print(torch_model)
```

Define Loss Function and Optimizer

```python
# Binary Cross-Entropy loss
criterion = nn.BCELoss()

# Adam optimizer with learning rate 0.01
optimizer = optim.Adam(
    torch_model.parameters(),
    lr=0.01
)

print("Loss function and optimizer defined successfully!")
```

Train the PyTorch MLP

```python
epochs = 1000

for epoch in range(epochs):
    # Forward pass
    outputs = torch_model(X_torch)

    # Calculate loss
    loss = criterion(outputs, y_torch)

    # Clear previous gradients
    optimizer.zero_grad()

    # Backpropagation
    loss.backward()

    # Update weights
    optimizer.step()

print("PyTorch model training completed!")
```

PyTorch Predictions

```python
# Make predictions
with torch.no_grad():
    torch_predictions = torch_model(X_torch)

# Convert probabilities to classes
torch_classes = (torch_predictions >= 0.5).int()

print("Predicted probabilities:")
print(torch_predictions)

print("\nPredicted classes:")
print(torch_classes)

print("\nActual outputs:")
print(y_torch.int())
```

```python
correct = (torch_classes == y_torch.int()).sum().item()
total = y_torch.size(0)

torch_accuracy = correct / total

print("PyTorch Accuracy:", torch_accuracy)
print("Accuracy Percentage:", torch_accuracy * 100, "%")
```

TensorFlow Low-Level API

```python
import tensorflow as tf

# Create TensorFlow tensors
X_tf = tf.constant(X, dtype=tf.float32)
y_tf = tf.constant(y, dtype=tf.float32)

# Initialize weights and biases
W1 = tf.Variable(tf.random.normal([2, 4]))
b1 = tf.Variable(tf.zeros([4]))

W2 = tf.Variable(tf.random.normal([4, 1]))
b2 = tf.Variable(tf.zeros([1]))

print("TensorFlow low-level model initialized successfully!")
```

Define Forward Pass and Optimizer

```python
# Adam optimizer
optimizer_tf = tf.optimizers.Adam(learning_rate=0.01)

# Binary Cross-Entropy loss
loss_fn = tf.keras.losses.BinaryCrossentropy()

def forward_pass(X):
    # Hidden layer
    hidden = tf.tanh(tf.matmul(X, W1) + b1)

    # Output layer
    output = tf.sigmoid(tf.matmul(hidden, W2) + b2)

    return output

print("Forward pass and optimizer defined successfully!")
```

Train the TensorFlow Low-Level Model

```python
epochs = 1000

for epoch in range(epochs):
    with tf.GradientTape() as tape:
        predictions_tf = forward_pass(X_tf)
        loss = loss_fn(y_tf, predictions_tf)

    gradients = tape.gradient(
        loss,
        [W1, b1, W2, b2]
    )

    optimizer_tf.apply_gradients(
        zip(gradients, [W1, b1, W2, b2])
    )

print("TensorFlow low-level model training completed!")
print("Final Loss:", float(loss))
```

Make TensorFlow Predictions

```python
# Generate predictions
predictions_tf = forward_pass(X_tf)

# Convert probabilities to binary classes
classes_tf = tf.cast(predictions_tf >= 0.5, tf.int32)

print("Predicted probabilities:")
print(predictions_tf.numpy())

print("\nPredicted classes:")
print(classes_tf.numpy())

print("\nActual outputs:")
print(y_tf.numpy().astype(int))
```

```python
correct_tf = tf.reduce_sum(
    tf.cast(classes_tf == tf.cast(y_tf, tf.int32), tf.int32)
)

total_tf = tf.size(y_tf)

tf_accuracy = float(correct_tf.numpy() / total_tf.numpy())

print("TensorFlow Low-Level Accuracy:", tf_accuracy)
print("Accuracy Percentage:", tf_accuracy * 100, "%")
```

Experiment with Number of Neurons

```python
from tensorflow import keras
from tensorflow.keras import layers

neurons_list = [2, 4, 8]

for neurons in neurons_list:
    test_model = keras.Sequential([
        layers.Input(shape=(2,)),
        layers.Dense(neurons, activation='tanh'),
        layers.Dense(1, activation='sigmoid')
    ])

    test_model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.01),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    test_model.fit(X, y, epochs=1000, verbose=0)

    loss, accuracy = test_model.evaluate(X, y, verbose=0)

    print(f"Neurons: {neurons} | Accuracy: {accuracy * 100:.2f}%")
```

```python
learning_rates = [0.001, 0.01, 0.1]

for lr in learning_rates:
    test_model = keras.Sequential([
        layers.Input(shape=(2,)),
        layers.Dense(4, activation='tanh'),
        layers.Dense(1, activation='sigmoid')
    ])

    test_model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    test_model.fit(X, y, epochs=1000, verbose=0)

    loss, accuracy = test_model.evaluate(X, y, verbose=0)

    print(f"Learning Rate: {lr} | Accuracy: {accuracy * 100:.2f}% | Loss: {loss:.6f}")
```

Experiment with Activation Function

```python
activation_functions = ['tanh', 'relu']

for activation in activation_functions:
    test_model = keras.Sequential([
        layers.Input(shape=(2,)),
        layers.Dense(4, activation=activation),
        layers.Dense(1, activation='sigmoid')
    ])

    test_model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.01),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    test_model.fit(X, y, epochs=1000, verbose=0)

    loss, accuracy = test_model.evaluate(X, y, verbose=0)

    print(
        f"Activation: {activation} | "
        f"Accuracy: {accuracy * 100:.2f}% | "
        f"Loss: {loss:.6f}"
    )
```
